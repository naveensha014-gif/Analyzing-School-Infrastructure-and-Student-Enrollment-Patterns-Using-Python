# **Stage 2 ** -  Data Cleaning and Pre-processing

```python
# Handle duplicates

duplicates = df.duplicated().sum()
print("Duplicate Rows:", duplicates)
```

# Observation
Duplicate Rows Found: 0

No duplicate records are present in the dataset.

```python
# Handle missing values (impute or drop)

missing_values = df.isnull().sum().sort_values(ascending=False)
print(missing_values)
```

```python
# Drop Highly Missing Columns
# Columns with more than 70% missing values has bene removed because they add very little analytical value.

df.drop(columns=['aff_board_h_sec', 'aff_board_sec', 'ward'],inplace=True)
```

```python
# Fill Missing Numerical Values

num_cols = ['district_code', 'subdistrict_code', 'pincode']
for col in num_cols:
  df[col] = df[col].fillna(df[col].median())

```

```python
# Fill Missing Categorical Values

df['subdistrict_name'] = df['subdistrict_name'].fillna(df['subdistrict_name'].mode()[0])
```

```python
# Renaming columns improves readability and usability.

df.rename(columns={
    'state_name' : 'State',
    'district_name' : 'District',
    'school_name' : 'School_Name',
    'management' : 'Management_Type',
    'total_teachers' : 'Total_Teachers',
    'class_students' : 'Total_Students'
}, inplace= True)
```

```python
# Data Type Conversion
# Converting object type school code into string:

df['udise_school_code'] = df['udise_school_code'].astype(str)

# Converting numeric columns:

numeric_cols = ['class_from', 'class_to', 'Total_Teachers', 'Total_Students']
for col in numeric_cols:
  df[col] = pd.to_numeric(df[col], errors='coerce')

```

```python
from numpy._core.defchararray import upper
# Outlier Treatment
# Outliers may exist in: Total students, Total teachers, Classroom counts
# Detecting Outliers using IQR Method:

Q1 = df['Total_Students'].quantile(0.25)
Q3 = df['Total_Students'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q1 + 1.5 * IQR

df = df[(df['Total_Students'] >= lower_bound) &
        (df['Total_Students'] <= upper_bound)]

```

## Data Transformation / Data Wrangling

```python

# Total Students Per School

df.loc[:,'Student_Teacher_Ratio'] = (df['Total_Students'] / df['Total_Teachers'])
```

```python
# # Calculating school age;

current_year = 2026

df.loc[:,'School_Age']= current_year - df['year_of_establishment']
```

```python
# # Categorizing school size:

def school_size(x):
  if x <100:
    return 'Small'
  elif x <500:
    return 'Medium'
  else:
    return 'Large'

```

# **Stage 3 : Statistical analysis & Visualizations****


```python
# Measure of Central Tendency
#  Mean, Median, and Mode for Total Students

# Mean
mean_students = df['Total_Students'].mean()

# Median
median_students = df['Total_Students'].median()

# Mode
mode_students = df['Total_Students'].mode()[0]

print("Mean:", mean_students)
print("Median:", median_students)
print("Mode:", mode_students)
```

```python
mean_students = df['Total_Students'].mean()
median_students = df['Total_Students'].median()
mode_students = df['Total_Students'].mode()[0]

# Plot
plt.figure(figsize=(10,6))
sns.histplot(df['Total_Students'], bins=20, kde=True)

# Add Mean, Median, Mode lines
plt.axvline(mean_students, color='red', linestyle='--',
            label=f'Mean = {mean_students:.2f}')
plt.axvline(median_students, color='green', linestyle='--',
            label=f'Median = {median_students:.2f}')
plt.axvline(mode_students, color='blue', linestyle='--',
            label=f'Mode = {mode_students:.2f}')

plt.title('Distribution of Total Students')
plt.xlabel('Total Students')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
